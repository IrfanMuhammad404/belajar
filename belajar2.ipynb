```python
import pandas as pd

df = pd.read_csv('drive/MyDrive/learn/drug200.csv')
df.describe(include=('all'))

```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder

#salin dataframe untuk proses encoding
df_salin = df.copy()
df_salin
```

```python
# Mengambil ulang kolom asli dari df untuk mencegah NaN jika sel dijalankan ulang
df_salin['BP'] = df['BP']
df_salin['Cholesterol'] = df['Cholesterol']

# Untuk tipe data ordinal lakukan mapping dengan membuat dictionary yang baru
bp_mapping = {'LOW': 0, 'NORMAL': 1, 'HIGH': 2}
chol_mapping = {'NORMAL': 0, 'HIGH': 1}

# Gunakan .map untuk memasukkan nilai mapping ke dalam variabel
df_salin['BP'] = df_salin['BP'].map(bp_mapping)
df_salin['Cholesterol'] = df_salin['Cholesterol'].map(chol_mapping)

display(df_salin.head())
```

```python
# Menginisialisasi LabelEncoder
le_sex = LabelEncoder()
le_drug = LabelEncoder()
le_sex
```

```python

# Melakukan encoding pada kolom Sex dan Drug
df_salin['Sex'] = le_sex.fit_transform(df_salin['Sex'])
df_salin['Drug'] = le_drug.fit_transform(df_salin['Drug'])

# Menampilkan hasil mapping encoder
print("Mapping Sex:", dict(zip(le_sex.classes_, le_sex.transform(le_sex.classes_))))
print("Mapping Drug:", dict(zip(le_drug.classes_, le_drug.transform(le_drug.classes_))))

# Menampilkan data teratas hasil encoding
display(df_salin.head())
```
