```python
# Preopcessing menggunaakan sklearn.preprocessing
from sklearn.preprocessing import LabelEncoder as LE
df_encode = df.copy()
df_encode.head()
```

```python
# inisialisasi encoder untuk tiap variabel yang ingin di encode
le_car_name = LE()
le_fuel_type = LE()
le_seller_type = LE()
le_transmission = LE()
```

```python
#masukkan label encode kedalam variabel
df_encode['Car_Name'] = le_car_name.fit_transform(df_encode['Car_Name'])
df_encode['Fuel_Type'] = le_fuel_type.fit_transform(df_encode['Fuel_Type'])
df_encode['Seller_Type'] = le_seller_type.fit_transform(df_encode['Seller_Type'])
df_encode['Transmission'] = le_transmission.fit_transform(df_encode['Transmission'])

df_encode.head()
```

```python
# memisahkan variabel target
y = df_encode['Selling_Price']
y
```

```python
# memisahkan varibael fitur
fitur = ['Car_Name',"Year", 'Present_Price','Kms_Driven','Fuel_Type','Seller_Type','Transmission','Owner']
x = df_encode[fitur]
x
```

```python
# Mmembuat model deccision tree

from sklearn.tree import DecisionTreeRegressor as DTR
df_model = DTR()
df_model.fit(x,y)
```

```python
# melakukan prediksi dengan model

df_model.predict(x.head())
```

```python
# melihat nilai mean absolute eror
from sklearn.metrics import mean_absolute_error as MAE
nilai_mae = MAE(y, df_model.predict(x))
nilai_mae
```

```python
from sklearn.model_selection import train_test_split as TTS
train_x, val_x, train_y, val_y = TTS(x,y, random_state=0)
train_y
```

```python
from sklearn.tree import DecisionTreeRegressor as DTR
from sklearn.metrics import mean_absolute_error as MAE
train_model = DTR()
train_model.fit(train_x,train_y)

prediksi = train_model.predict(val_x)
prediksi
```

```python
nilai_mae = MAE(val_y, prediksi)
nilai_mae
```

```python
from sklearn.ensemble import RandomForestRegressor as RFR
from sklearn.metrics import mean_absolute_error as MAE

forest_model = RFR(random_state=1)
forest_model.fit(train_x,train_y)

forest_model.predict(val_x)
mae_random_forest = MAE(val_y, forest_model.predict(val_x))
mae_random_forest
```
