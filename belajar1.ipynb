```python
import pandas as pd
# Pandas merupakan perangkat utama untuk mengeksplorasi dan memanipulasi data
```

```python
melb = 'drive/MyDrive/learn/melb_data.csv'
df = pd.read_csv(melb)
df.describe()


# pd.read_csv dan langsung koneksikan dengan directory gdrive untuk langsung mendeskirpsikan dataset
```

```python
#menampilkan smua nama kolom dalam dataset df
df.columns
```

```python
#melihat nilai null
df.isnull().sum()
```

```python
# menghilangkan nilai NA dari dataset
df = df.dropna(axis=0)
df.isnull().sum()
```

```python
# memisahkan y sebagai nilai target
y = df.Price
y
```

```python
# memilih dan membuat fitur menjadi satu kesatuan dalam variabel data frame x
fitur = ['Rooms', 'Bathroom', 'Landsize', 'BuildingArea',
                        'YearBuilt', 'Lattitude', 'Longtitude']
x = df[fitur]
x.describe()
```

```python
x.head()
```

```python
# Membuat model decision tree dengan sklearn.tree

from sklearn.tree import DecisionTreeRegressor

df_model = DecisionTreeRegressor(random_state=1) # random_state = 1 untuk menjaga konsistensi hasil pada setiap eksekusi
df_model.fit(x, y)
df_model
```

```python
# uji coba model dengan memprediksi harga pada r baris pertama
print ('Kita akan membuat prediksi harga rumah berikut: ')
print (x.head())
print ('Prediksi harga rumah: ')
print (df_model.predict(x.head()))
```

```python
# Menentukan mean absolute error

from sklearn.metrics import mean_absolute_error

prediksi_harga_rumah = df_model.predict(x)
mean_absolute_error(y, prediksi_harga_rumah)

```

```python
# Memisahkan data untuk memprediksi data train dan test

from sklearn.model_selection import train_test_split

train_x, val_x, train_y, val_y = train_test_split(x, y, random_state = 0)
```

```python
df = DecisionTreeRegressor()
df.fit(train_x, train_y)
```

```python
prediksi_harga_rumah = df.predict(val_x)
print('Mean Absolute Error pada data validasi : ')
print(mean_absolute_error(val_y, prediksi_harga_rumah))
```

```python
prediksi_harga_rumah1 = df.predict(train_x)
print('Mean Absolute Error pada data train')
print(mean_absolute_error(train_y, prediksi_harga_rumah1))
```
